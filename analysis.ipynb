```python
import numpy as np
from matplotlib import pyplot as plt

#data = np.fromfile('data_gil_500.bin', dtype = np.float64).reshape(6,-1)
data = np.fromfile("Canard_gil_3D.bin", dtype=np.float64).reshape(3,-1)
#data = np.fromfile('data_gil_100000.bin', dtype = np.float64).reshape(6,-1)
omega=10000
V0 = data[0]
V1 = data[1]
V2 = 1-np.array(V0)-np.array(V1)
RT = data[2]
```

```python
plt.style.use('/Users/lhc642/.config/matplotlib/stylelib/PRE_1col.mplstyle')
fig, ax = plt.subplots(1,1,figsize=(3.75,1.))
ax2 = ax.twinx()
lim = 6000
ax.plot(RT[:lim], np.log(V0[:lim]), 'darkorange',lw=0.75) 
ax.plot(RT[:lim], np.log(V1[:lim]), 'forestgreen',lw=0.75)
ax.plot(RT[:lim], np.log(V2[:lim]), 'cornflowerblue',lw=0.75)

V_max= np.max((V0[:lim]))
ax2.axhline(y=np.log(V_max), color='black', linestyle='--', linewidth=0.25)
ax2.axhline(y=np.log(V_max*0.1391**1), color='black', linestyle='--', linewidth=0.25)
ax2.axhline(y=np.log(V_max*0.1391**2), color='black', linestyle='--', linewidth=0.25)
ax2.axhline(y=np.log(V_max*0.1391**3), color='black', linestyle='--', linewidth=0.25)

ax2.spines['right'].set_visible(True)   
ax.set_xlabel('Time', )
ax.set_ylabel('$\log(V_i)$', )
ax.set_xticks([0, 100, 200,])
ax.set_yticks([0, -5,])
ax.set_ylim(-7, 0)
ax2.set_ylim(-7, 0)
ax2.set_ylabel('$m$', )
ax2.set_yticks([np.log(V_max), np.log(V_max*0.1391**1), np.log(V_max*0.1391**2), np.log(V_max*0.1391**3)])
ax2.set_yticklabels(['$0$', '$1$', '$2$', '$3$'])
plt.savefig('fig4.png', )
plt.show()
```
